```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("ticks")
```

```python
df = pd.read_parquet("../Analysis_Outputs/Group_Signal-Flow/Group_Functional-Roles_2026-04-03.parquet")
df
```

```python
linewidth      = 0.75
scatter_size   = 2
scatter_alpha  = 0.75
linewidth_hist = 0.5
alpha_hist     = 0.35
title_fontsize = 8.5
label_fontsize = 7
tick_fontsize  = 6
panel_fontsize = 16
```

```python
from IPython.core import display_trap
fig, ax = plt.subplots(figsize = (3, 6), nrows = 2, sharex = True)
plt.subplots_adjust(hspace = 0.325)

sns.stripplot(df,
              x = "Role",
              y = "w_out",
              jitter = True,
              ax = ax[0],
              s  = scatter_size,
              alpha = scatter_alpha)

sns.stripplot(df,
              x = "Role",
              y = "w_in",
              jitter = True,
              ax = ax[1],
              s  = scatter_size,
              alpha = scatter_alpha)

# Desping
sns.despine(ax = ax[0])
sns.despine(ax = ax[1], bottom = True, right = True, top = False)

# Format
for a in ax:
    # Ticks
    a.tick_params(
        axis      = "both",
        which     = "major",
        color     = "black",
        labelsize = tick_fontsize,
        length    = 1.5,
        width     = 0.5,
        pad       = 1.5)
    # Limit
    a.set_ylim(-200, 7000)

# Invert
ax[1].invert_yaxis()
ax[1].xaxis.tick_top()

# Rotate
labs = ax[1].get_xticklabels()
labs = [lab.get_text().replace("_", "\n") for lab in labs]
ax[1].set_xticklabels(labels = labs, rotation = 90)

# Labels
ax[0].set_ylabel("Output Synapses", fontsize = label_fontsize, labelpad = 10)
ax[1].set_ylabel("Input Synapses", fontsize = label_fontsize, labelpad = 10)
ax[1].set_xlabel("")

# Save
plt.savefig("../Analysis_Outputs/Group_Signal-Flow/Figure_05_Supplementary_Synapse-Counts-Flow.pdf",
            bbox_inches = "tight",
            dpi = 1200)
```
